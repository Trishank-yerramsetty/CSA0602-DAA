```python
#1Q Brute Force

import math

def closest_pair(points):
    min_distance = float('inf')
    closest_points = None

    n = len(points)

    # Check every pair of points
    for i in range(n):
        for j in range(i + 1, n):
            x1, y1 = points[i]
            x2, y2 = points[j]

            # Euclidean distance
            distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

            if distance < min_distance:
                min_distance = distance
                closest_points = (points[i], points[j])

    return closest_points, min_distance


# Input
points = [(1, 2), (4, 5), (7, 8), (3, 1)]

# Find closest pair
pair, distance = closest_pair(points)

print("Closest pair:", pair[0], "-", pair[1])
print("Minimum distance:", distance)
```

```python
#6Q Brute Force

# Convex Hull using Jarvis March (Gift Wrapping) Method

def orientation(p, q, r):
    return (q[1] - p[1]) * (r[0] - q[0]) - \
           (q[0] - p[0]) * (r[1] - q[1])

def convex_hull(points):
    n = len(points)

    if n < 3:
        return points

    hull = []

    # Find the leftmost point
    leftmost = 0
    for i in range(1, n):
        if points[i][0] < points[leftmost][0]:
            leftmost = i

    p = leftmost

    while True:
        hull.append(points[p])

        q = (p + 1) % n

        for r in range(n):
            if orientation(points[p], points[q], points[r]) < 0:
                q = r

        p = q

        if p == leftmost:
            break

    return hull


# Input
points = [(0, 0), (1, 1), (2, 2), (0, 2), (2, 0)]

# Find Convex Hull
hull = convex_hull(points)

print("Convex Hull Points:")
for point in hull:
    print(point)
```

```python
#8Q Brute Force

# Convex Hull using Basic (Brute Force) Method

def orientation(p, q, r):
    return (q[0] - p[0]) * (r[1] - p[1]) - \
           (q[1] - p[1]) * (r[0] - p[0])

def convex_hull(points):
    n = len(points)
    hull = set()

    for i in range(n):
        for j in range(i + 1, n):

            pos = neg = 0

            for k in range(n):
                if k == i or k == j:
                    continue

                val = orientation(points[i], points[j], points[k])

                if val > 0:
                    pos += 1
                elif val < 0:
                    neg += 1

            # Edge of convex hull if all points are on one side
            if pos == 0 or neg == 0:
                hull.add(points[i])
                hull.add(points[j])

    return list(hull)


# Input
points = [(0, 3), (1, 1), (2, 2), (4, 4), (3, 0)]

# Find Convex Hull
hull = convex_hull(points)

print("Convex Hull Points:")
for point in hull:
    print(point)
```

```python
#11Q Brute Force

from itertools import combinations

def subset_sum(arr, target):
    n = len(arr)

    # Check all possible subsets
    for r in range(1, n + 1):
        for subset in combinations(arr, r):
            if sum(subset) == target:
                return list(subset)

    return None


# Input
arr = [3, 34, 4, 12, 5, 2]
target = 9

# Find subset
result = subset_sum(arr, target)

if result:
    print("Subset found:", result)
    print("Sum:", sum(result))
else:
    print("No subset found")
```

```python
#12Q

import math
from itertools import permutations

# Function to calculate distance between two cities
def distance(city1, city2):
    return math.sqrt((city1[0] - city2[0])**2 +
                     (city1[1] - city2[1])**2)

# Function to find shortest path using exhaustive search
def tsp_bruteforce(cities):
    min_distance = float('inf')
    best_path = None

    start_city = cities[0]

    # Generate all possible paths
    for perm in permutations(cities[1:]):
        path = [start_city] + list(perm) + [start_city]

        total_distance = 0
        for i in range(len(path) - 1):
            total_distance += distance(path[i], path[i + 1])

        if total_distance < min_distance:
            min_distance = total_distance
            best_path = path

    return min_distance, best_path


# --------------------
# Test Case 1
# --------------------
cities1 = [(1, 2), (4, 5), (7, 1), (3, 6)]

dist1, path1 = tsp_bruteforce(cities1)

print("Test Case 1:")
print("Shortest Distance:", dist1)
print("Shortest Path:", path1)
# Test Case 2
cities2 = [(2, 4), (8, 1), (1, 7), (6, 3), (5, 9)]

dist2, path2 = tsp_bruteforce(cities2)

print("\nTest Case 2:")
print("Shortest Distance:", dist2)
print("Shortest Path:", path2)
```

```python
#13Q

from itertools import permutations

# Function to calculate total cost of an assignment
def total_cost(assignment, cost_matrix):
    cost = 0
    for worker in range(len(assignment)):
        task = assignment[worker]
        cost += cost_matrix[worker][task]
    return cost


# Function to solve Assignment Problem using Exhaustive Search
def assignment_problem(cost_matrix):
    n = len(cost_matrix)

    min_cost = float('inf')
    best_assignment = None

    # Generate all possible task assignments
    for assignment in permutations(range(n)):
        cost = total_cost(assignment, cost_matrix)

        if cost < min_cost:
            min_cost = cost
            best_assignment = assignment

    return best_assignment, min_cost


# -------------------------
# Test Case 1
# -------------------------
cost_matrix1 = [
    [3, 10, 7],
    [8, 5, 12],
    [4, 6, 9]
]

assignment1, cost1 = assignment_problem(cost_matrix1)

print("Test Case 1:")
print("Optimal Assignment:")
for worker, task in enumerate(assignment1, start=1):
    print(f"(worker {worker}, task {task + 1})")

print("Total Cost:", cost1)


# -------------------------
# Test Case 2
# -------------------------
cost_matrix2 = [
    [15, 9, 4],
    [8, 7, 18],
    [6, 12, 11]
]

assignment2, cost2 = assignment_problem(cost_matrix2)

print("\nTest Case 2:")
print("Optimal Assignment:")
for worker, task in enumerate(assignment2, start=1):
    print(f"(worker {worker}, task {task + 1})")

print("Total Cost:", cost2)
```

```python
#14Q

from itertools import combinations

# Function to calculate total value
def total_value(items, values):
    value = 0
    for item in items:
        value += values[item]
    return value


# Function to check feasibility
def is_feasible(items, weights, capacity):
    total_weight = 0

    for item in items:
        total_weight += weights[item]

    return total_weight <= capacity


# Function to solve 0-1 Knapsack using Exhaustive Search
def knapsack_exhaustive(weights, values, capacity):
    n = len(weights)

    best_selection = []
    max_value = 0

    # Generate all possible subsets
    for r in range(n + 1):
        for subset in combinations(range(n), r):

            if is_feasible(subset, weights, capacity):
                value = total_value(subset, values)

                if value > max_value:
                    max_value = value
                    best_selection = list(subset)

    return best_selection, max_value


# -------------------------
# Test Case 1
# -------------------------
weights1 = [2, 3, 1]
values1 = [4, 5, 3]
capacity1 = 4

selection1, value1 = knapsack_exhaustive(weights1, values1, capacity1)

print("Test Case 1:")
print("Optimal Selection:", selection1)
print("Total Value:", value1)


# -------------------------
# Test Case 2
# -------------------------
weights2 = [1, 2, 3, 4]
values2 = [2, 4, 6, 3]
capacity2 = 6

selection2, value2 = knapsack_exhaustive(weights2, values2, capacity2)

print("\nTest Case 2:")
print("Optimal Selection:", selection2)
print("Total Value:", value2)
```

```python
#15Q

from itertools import combinations

# Function to solve Subset Sum using Exhaustive Search
def subset_sum(arr, target):
    n = len(arr)

    # Generate all possible subsets
    for r in range(1, n + 1):
        for subset in combinations(arr, r):

            if sum(subset) == target:
                return list(subset)

    return None


# Input
arr = [15, 10, 12, 7, 5]
target = 22

# Find subset
result = subset_sum(arr, target)

if result:
    print("Subset found:", result)
    print("Sum:", sum(result))
else:
    print("No subset found")
```
