```python
#1Q Selection Sort

def top_k_scores(scores, k):
    # Create a copy so the original list is not modified
    arr = scores.copy()
    n = len(arr)

    # If k is greater than the number of elements
    k = min(k, n)

    # Selection sort approach: find maximum repeatedly
    for i in range(k):
        max_index = i

        # Find the maximum element in the remaining array
        for j in range(i + 1, n):
            if arr[j] > arr[max_index]:
                max_index = j

        # Swap the maximum element to the current position
        arr[i], arr[max_index] = arr[max_index], arr[i]

    # Return only the top k scores
    return arr[:k]
scores = [72, 88, 65, 90, 77, 95, 60, 83, 91, 68]
k = 5
print(top_k_scores(scores, k))
```

```python
#2Q Selection Sort

def selection_sort_min_writes(arr):
    a = arr.copy()  # Create a copy to preserve original list
    n = len(a)
    swap_count = 0

    # Selection Sort
    for i in range(n - 1):
        min_index = i

        # Find the minimum element in the unsorted part
        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j

        # Swap only if needed
        if min_index != i:
            a[i], a[min_index] = a[min_index], a[i]
            swap_count += 1

    return a, swap_count
data = [23.5, 19.2, 25.1, 18.8, 21.4]

sorted_data, swaps = selection_sort_min_writes(data)

print("Sorted Data:", sorted_data)
print("Number of Swaps:", swaps)
```

```python
#3Q Selection Sort
def reorder_shelf(books):
    # Create a copy so the original list is not modified
    arr = books.copy()
    n = len(arr)
    moves = 0

    # Selection Sort
    for i in range(n - 1):
        min_index = i

        # Find the smallest book ID in the remaining shelf
        for j in range(i + 1, n):
            if arr[j] < arr[min_index]:
                min_index = j

        # Perform a physical move (swap) only if needed
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            moves += 1

    return arr, moves
books = [305, 102, 250, 118, 199, 400, 101]

sorted_books, moves = reorder_shelf(books)

print("Reordered Shelf:", sorted_books)
print("Physical Moves:", moves)
```

```python
#4Q Selection Sort

def distribute_prizes(participants):
    # Create a copy so the original list is not modified
    arr = participants.copy()
    n = len(arr)

    # Selection Sort in Descending Order
    for i in range(n - 1):
        max_index = i

        # Find participant with highest score
        for j in range(i + 1, n):
            if arr[j][1] > arr[max_index][1]:
                max_index = j

        # Swap if needed
        if max_index != i:
            arr[i], arr[max_index] = arr[max_index], arr[i]

    # Print ranking
    print("Contest Ranking:")
    for rank, participant in enumerate(arr, start=1):
        print(f"{rank}. {participant[0]} - {participant[1]}")

    return arr


# Test Cases
ranking = distribute_prizes([
    ('Asha', 88),
    ('Ravi', 95),
    ('Meera', 79),
    ('Dev', 95)
])

scores_only = [p[1] for p in ranking]

assert scores_only == sorted(scores_only, reverse=True)  # descending order
assert ranking[0][1] == 95  # top score first

print("All test cases passed!")
scores_only = [p[1] for p in ranking]

assert scores_only == sorted(scores_only, reverse=True)  # descending order
assert ranking[0][1] == 95  # top score first

print("All test cases passed!")
```

```python
#5Q Selection Sort

import time

def selection_sort(arr):
    # Create a copy so the original list is not modified
    a = arr.copy()
    n = len(a)

    # Selection Sort
    for i in range(n - 1):
        min_index = i

        # Find the minimum element
        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j

        # Swap if needed
        if min_index != i:
            a[i], a[min_index] = a[min_index], a[i]

    return a

# Timing the sort
prices = [499, 129, 899, 45, 275, 60, 310, 150]

start_time = time.perf_counter()
sorted_prices = selection_sort(prices)
end_time = time.perf_counter()

print("Original Prices:", prices)
print("Sorted Prices:", sorted_prices)
print("Execution Time:", (end_time - start_time) * 1000000, "microseconds")


# Test Cases
assert selection_sort([499, 129, 899, 45, 275, 60, 310, 150]) == \
       sorted([499, 129, 899, 45, 275, 60, 310, 150])

assert selection_sort([]) == []      # empty list
assert selection_sort([7]) == [7]    # single item

print("All test cases passed!")
```

```python
def optimized_bubble_sort(arr):
    a = arr.copy()  # Create a copy of the list
    n = len(a)
    passes = 0

    for i in range(n - 1):
        swapped = False
        passes += 1

        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Early exit if no swaps occurred
        if not swapped:
            break

    return a, passes


# Test Cases
sorted_rolls, passes = optimized_bubble_sort(
    [101, 102, 104, 103, 105, 107, 106, 108]
)
roll_numbers = [101, 102, 104, 103, 105, 107, 106, 108]

sorted_rolls, passes = optimized_bubble_sort(roll_numbers)

print("Sorted Roll Numbers:", sorted_rolls)
print("Number of Passes:", passes)
```

```python
#2Q Bubble Sort

def bubble_sort_queue(queue):
    # Priority values
    priority = {
        "ambulance": 3,
        "bus": 2,
        "car": 1
    }

    arr = queue.copy()
    n = len(arr)

    # Bubble Sort (Descending Priority)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if priority[arr[j]] < priority[arr[j + 1]]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    return arr
queue = ['car', 'car', 'bus']

print("Before Arrival:", queue)

queue.append('ambulance')

sorted_queue = bubble_sort_queue(queue)

print("After Arrival:", sorted_queue)
```

```python
#3Q Bubble Sort

def bubble_sort_with_frames(arr):
    a = arr.copy()
    n = len(a)

    # Store the initial state as the first frame
    frames = [a.copy()]

    # Bubble Sort
    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Store array state after each pass
        frames.append(a.copy())

        # Early exit if already sorted
        if not swapped:
            break

    return frames
frames = bubble_sort_with_frames([5, 1, 4, 2, 8])

for i, frame in enumerate(frames):
    print(f"Pass {i}: {frame}")
```

```python
#4Q Bubble Sort

def bubble_sort_plain(arr):
    a = arr.copy()
    n = len(a)
    comparisons = 0

    for i in range(n - 1):
        for j in range(n - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]

    return a, comparisons


def bubble_sort_optimized(arr):
    a = arr.copy()
    n = len(a)
    comparisons = 0

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Early termination if no swaps occur
        if not swapped:
            break

    return a, comparisons
alerts = [2, 1, 3, 2, 1, 4, 3, 2, 5, 1, 2, 3, 4, 1, 2]

sorted_alerts, comparisons = bubble_sort_optimized(alerts)

print("Sorted Alerts:", sorted_alerts)
print("Comparisons:", comparisons)
```

```python
#5Q Bubble Sort

import random

def bubble_sort_hand(hand):
    a = hand.copy()
    n = len(a)
    passes = 0

    for i in range(n - 1):
        swapped = False
        passes += 1

        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        # Early exit if already sorted
        if not swapped:
            break

    return a, passes

hand = [2, 4, 6, 8, 9, 11, 13]
hand.append(7)

sorted_hand, passes_incremental = bubble_sort_hand(hand)

print("Nearly Sorted Hand:", sorted_hand)
print("Passes Needed:", passes_incremental)

shuffled = hand.copy()
random.shuffle(shuffled)

sorted_shuffled, passes_full = bubble_sort_hand(shuffled)

print("\nShuffled Hand:", shuffled)
print("Sorted Hand:", sorted_shuffled)
print("Passes Needed:", passes_full)
```

```python
#1Q Insertion Sort

def insert_updated_score(board, new_score):
    # Create a copy of the leaderboard
    arr = board.copy()

    # Add the updated score at the end
    arr.append(new_score)

    shifts = 0
    i = len(arr) - 2

    # Insert in descending order (highest score first)
    while i >= 0 and arr[i] < new_score:
        arr[i + 1] = arr[i]
        shifts += 1
        i -= 1

    arr[i + 1] = new_score

    return arr, shifts
board = [980, 875, 760, 690, 500]

updated_board, shifts = insert_updated_score(board, 820)

print("Updated Leaderboard:", updated_board)
print("Number of Shifts:", shifts)
```

```python
#2Q Insertion Sort

def pick_up_card(hand, card):
    # Create a copy of the current hand
    arr = hand.copy()

    # Add the new card at the end
    arr.append(card)

    # Insertion Sort logic
    i = len(arr) - 2

    while i >= 0 and arr[i] > card:
        arr[i + 1] = arr[i]
        i -= 1

    arr[i + 1] = card

    return arr
hand = []

for card in [7, 2, 9, 4, 1]:
    hand = pick_up_card(hand, card)
    print("After picking up", card, ":", hand)
```

```python
#3Q Insertion Sort

def insert_price(prices, new_price):
    # Create a copy of the sorted price list
    arr = prices.copy()

    # Add the new price at the end
    arr.append(new_price)

    # Insertion Sort logic
    i = len(arr) - 2

    while i >= 0 and arr[i] > new_price:
        arr[i + 1] = arr[i]
        i -= 1

    arr[i + 1] = new_price

    return arr
prices = []

for p in [102.5, 98.3, 105.1, 100.0, 97.8]:
    prices = insert_price(prices, p)
    print("After inserting", p, ":", prices)
```

```python
#4Q Insertion Sort

def insert_song(playlist, new_song):
    # Create a copy of the playlist
    arr = playlist.copy()

    # Add the new song at the end
    arr.append(new_song)

    # Insertion Sort logic
    i = len(arr) - 2

    while i >= 0 and arr[i][1] > new_song[1]:
        arr[i + 1] = arr[i]
        i -= 1

    arr[i + 1] = new_song

    return arr
playlist = [
    ('Intro', 120),
    ('Chill Beat', 210),
    ('Long Jam', 340)
]

updated_playlist = insert_song(playlist, ('Quick Track', 180))

print("Updated Playlist:")
for song in updated_playlist:
    print(song)
```

```python
#5Q Insertion Sort

import random

def insertion_sort_count_shifts(arr):
    a = arr.copy()
    shifts = 0

    # Insertion Sort
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            shifts += 1
            j -= 1

        a[j + 1] = key

    return a, shifts
log = [18.2, 18.5, 18.9, 17.9, 19.1, 19.4, 19.0]

sorted_log, shifts_nearly = insertion_sort_count_shifts(log)

print("Original Log:", log)
print("Sorted Log:", sorted_log)
print("Shifts (Nearly Sorted):", shifts_nearly)
```
