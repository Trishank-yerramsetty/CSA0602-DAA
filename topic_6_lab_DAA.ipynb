```python
#1Q

def merge_sort(arr):
    comparisons = 0

    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2

    left, c1 = merge_sort(arr[:mid])
    right, c2 = merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        comparisons += 1

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result += left[i:]
    result += right[j:]

    comparisons += c1 + c2

    return result, comparisons


arr = [12, 4, 78, 23, 45, 67, 89, 1]

sorted_arr, comparisons = merge_sort(arr)

print("Sorted Array:", sorted_arr)
print("Comparisons:", comparisons)
```

```python
#6Q

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result += left[i:]
    result += right[j:]

    return result


def hybrid_merge_sort(arr, threshold=4):

    if len(arr) <= threshold:
        return insertion_sort(arr)

    mid = len(arr) // 2

    left = hybrid_merge_sort(arr[:mid], threshold)
    right = hybrid_merge_sort(arr[mid:], threshold)

    return merge(left, right)


arr = [12, 11, 13, 5, 6, 7, 3, 2]

result = hybrid_merge_sort(arr)

print("Sorted Array:", result)
```

```python
#7Q

def merge_sort(arr):
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2

    left, count1 = merge_sort(arr[:mid])
    right, count2 = merge_sort(arr[mid:])

    result = []
    i = 0
    j = 0
    count = count1 + count2

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            count += len(left) - i
            j += 1

    result += left[i:]
    result += right[j:]

    return result, count


arr = [2, 4, 1, 3, 5]

sorted_arr, inversions = merge_sort(arr)

print("Sorted:", sorted_arr)
print("Inversions:", inversions)
```

```python
#11Q

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def kth_smallest(arr, low, high, k):

    if low <= high:

        position = partition(arr, low, high)

        if position == k - 1:
            return arr[position]

        elif position > k - 1:
            return kth_smallest(arr, low, position - 1, k)

        else:
            return kth_smallest(arr, position + 1, high, k)


arr = [7, 10, 4, 3, 20, 15]
k = 3

result = kth_smallest(arr, 0, len(arr) - 1, k)

print("K-th smallest element:", result)
```
